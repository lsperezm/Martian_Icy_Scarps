```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## South 1

```python
lats_S1= np.array([-56.62,-56.611,-56.669,-56.663])
longs_S1=np.array([113.981,114.096,113.992,114.107])
elevs_S1=np.array([2718.612,2829.382,2403.596,2733.016])
```

```python
def dip_dipdirect (lati, long, elev):
    R = 3389500 #Mars radius

    lon = np.radians(long)
    lat = np.radians(lati)

    # Use the center of the area as the origin
    lon0 = np.mean(lon)
    lat0 = np.mean(lat)

    # Equirectangular approximation 
    x = R * (lon - lon0) * np.cos(lat0)
    y = R * (lat - lat0)

    z = elev

    # Fit the plane
    A = np.column_stack((x, y, np.ones_like(x))) #Matrix

    # Least-squares solution to find coefficients
    a, b, c = np.linalg.lstsq(A, z, rcond=None)[0]

    #Find slope
    dip = np.degrees(np.arctan(np.sqrt(a**2 + b**2)))
    print(f"Dip = {dip:.2f}°")
    #Find dip direction
    dip_dir = (np.degrees(np.arctan2(-a, -b)) + 360) % 360
    print(f"Dip direction = {dip_dir:.2f}°")

    return dip,dip_dir
```

```python
dip_S1=dip_dipdirect(lats_S1,longs_S1,elevs_S1)
```

```python
winddir=np.array([124.09,168.53,230.94,174.83,255.73,242.29,252.08,185.69,240.85])
wind= np.array([1.54,2.33,4.73, 2.97,3.33,5.13,3.54,3.73,4.45])
dips=np.array([dip_S1[0],dip_S2[0],dip_S3[0],dip_S4[0],dip_S5[0],dip_S6[0],dip_S7[0],dip_S13[0],dip_S19[0]])
dipdirs=np.array([dip_S1[1],dip_S2[1],dip_S3[1],dip_S4[1],dip_S5[1],dip_S6[1],dip_S7[1],dip_S13[1],dip_S19[1]])
extension=np.array([2.72,2.43,2.17,1.14,2.79,1.42,1.49,2.63,1.78])
#volumes=np.array([])
winddip_angle=np.abs(dipdirs-winddir)
winddip_angle
```
