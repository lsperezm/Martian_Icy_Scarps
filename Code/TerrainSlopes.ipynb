```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## South 1

```python
lats_S1= np.array([-56.62,-56.611,-56.669,-56.663])
longs_S1=np.array([113.981,114.096,113.992,114.107])
elevs_S1=np.array([2718.612,2829.382,2403.596,2733.016])
```

```python
def dip_dipdirect (lati, long, elev):
    R = 3389500 #Mars radius

    lon = np.radians(long)
    lat = np.radians(lati)

    # Use the center of the area as the origin
    lon0 = np.mean(lon)
    lat0 = np.mean(lat)

    # Equirectangular approximation 
    x = R * (lon - lon0) * np.cos(lat0)
    y = R * (lat - lat0)

    z = elev

    # Fit the plane
    A = np.column_stack((x, y, np.ones_like(x))) #Matrix

    # Least-squares solution to find coefficients
    a, b, c = np.linalg.lstsq(A, z, rcond=None)[0]

    #Find slope
    dip = np.degrees(np.arctan(np.sqrt(a**2 + b**2)))
    print(f"Dip = {dip:.2f}°")
    #Find dip direction
    dip_dir = (np.degrees(np.arctan2(-a, -b)) + 360) % 360
    print(f"Dip direction = {dip_dir:.2f}°")

    return
```

```python
dip_dipdirect(lats_S1,longs_S1,elevs_S1)
```
